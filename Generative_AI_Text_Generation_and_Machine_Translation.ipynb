Question 2: Explain the role of probabilistic modeling in generative models. How do
these models differ from discriminative models?
  - Probabilistic modeling plays a central role in generative models because it allows these models to learn and represent the probability distribution of data. In simple terms, probabilistic modeling helps generative models understand how likely certain data patterns are and how different features relate to each other. These models learn the **joint probability distribution**, denoted as P(X, Y) or P(X), where X represents the input data and Y represents labels if available. By learning this distribution, generative models can generate new, realistic data samples that resemble the original training data. For example, a generative model trained on images of faces learns the probability patterns of facial features and can generate entirely new faces that look realistic. Models such as Generative Adversarial Networks (GANs), Variational Autoencoders (VAEs), and Naive Bayes use probabilistic approaches to create new data or simulate real-world scenarios. Probabilistic modeling also helps handle uncertainty and variability in data, making generative models useful in applications like image generation, speech synthesis, and text generation.

    Generative models differ from discriminative models mainly in their objective and approach. Generative models focus on learning how the data is generated by modeling the joint probability distribution P(X, Y), which allows them to both generate new data and perform classification. In contrast, discriminative models focus only on learning the conditional probability P(Y|X), which means they learn how to distinguish between different classes rather than generate new data. Discriminative models are mainly used for prediction and classification tasks, such as identifying whether an email is spam or not. Examples of discriminative models include Logistic Regression, Support Vector Machines (SVM), and Neural Networks used for classification. In summary, generative models learn how data is created and can generate new samples, while discriminative models learn how to classify or predict labels based on existing data.


Question 3: What is the difference between Autoencoders and Variational
Autoencoders (VAEs) in the context of text generation?
  - Autoencoders and Variational Autoencoders (VAEs) are both neural network models used to learn efficient representations of data, but they differ significantly in their approach and capability, especially in the context of text generation. An Autoencoder consists of two main parts: an encoder and a decoder. The encoder compresses the input text into a fixed-size vector called the latent representation, and the decoder reconstructs the original text from this representation. The main goal of a basic Autoencoder is to minimize reconstruction error, meaning it tries to reproduce the input as accurately as possible. However, the latent space learned by a standard Autoencoder is not structured or continuous, which makes it difficult to generate new and meaningful text. It mainly works well for tasks such as text compression, feature extraction, and noise removal, but it is not ideal for generating new text samples.

    On the other hand, a Variational Autoencoder (VAE) is a probabilistic version of the Autoencoder that learns a structured and continuous latent space. Instead of encoding input text into a single fixed vector, a VAE encodes it into a probability distribution defined by a mean and variance. This allows the model to sample new points from the latent space and generate new, diverse text. VAEs introduce randomness during training, which helps them learn smoother and more meaningful latent representations. This structured latent space makes VAEs more suitable for text generation tasks, such as creating new sentences, paraphrasing text, and generating variations of existing content. In summary, while Autoencoders focus on reconstructing input text, VAEs focus on learning probability distributions that allow the generation of new and diverse text, making them more powerful for generative applications.


Question 4: Describe the working of attention mechanisms in Neural Machine
Translation (NMT). Why are they critical?
  - Attention mechanisms play a crucial role in Neural Machine Translation (NMT) by helping the model focus on the most relevant parts of the input sentence while generating each word of the translated sentence. In traditional encoder–decoder models without attention, the encoder converts the entire input sentence into a single fixed-length vector, which the decoder uses to generate the translation. However, this approach has limitations, especially for long sentences, because important information can be lost during compression. Attention mechanisms solve this problem by allowing the decoder to access all the encoder’s hidden states instead of relying on just one fixed vector.

    The attention mechanism works by assigning different weights, called attention scores, to each word in the input sentence based on its relevance to the current word being generated in the output sentence. During translation, for each output word, the decoder calculates attention scores between its current state and all encoder hidden states. These scores are converted into probabilities using a softmax function, and a weighted sum of the encoder hidden states is computed to create a context vector. This context vector contains the most relevant information needed to generate the next word in the translated sentence. As a result, the model can focus more on important words and less on irrelevant ones, improving translation accuracy.

    Attention mechanisms are critical because they improve the performance and efficiency of NMT models, especially when handling long and complex sentences. They help preserve important contextual information, improve alignment between source and target languages, and produce more accurate and natural translations. Attention also makes the model more flexible and interpretable, as it shows which input words the model focuses on during translation. Overall, attention mechanisms significantly enhance the quality and effectiveness of Neural Machine Translation systems.


Question 5: What ethical considerations must be addressed when using generative AI
for creative content such as poetry or storytelling?
  - When using generative AI for creative content such as poetry or storytelling, several important ethical considerations must be addressed to ensure responsible and fair use. One major concern is **authorship and originality**. Since generative AI learns from large datasets that include human-created works, there is a risk that the generated content may closely resemble or unintentionally copy existing poems or stories. This raises questions about intellectual property rights and who should be credited as the creator—the AI developer, the user, or the original authors whose work was used for training. Ensuring proper attribution and avoiding plagiarism is essential.

    Another important consideration is **bias and fairness**. Generative AI models may learn biases present in the training data, which can result in content that reflects stereotypes, discrimination, or offensive themes. This can negatively impact individuals or groups and spread harmful narratives. Therefore, it is important to monitor and filter AI-generated content to ensure it is respectful, inclusive, and unbiased. **Transparency** is also essential, meaning users and audiences should be informed when content is generated by AI rather than humans. This helps maintain trust and prevents deception.

    Additionally, there are concerns related to **misuse and misinformation**. Generative AI can be used to create misleading stories, fake news, or impersonate real individuals, which can harm reputations and spread false information. There is also the issue of **job displacement**, as AI-generated creative content may reduce opportunities for human writers and artists. Finally, respecting **privacy and data protection** is important, as AI systems should not generate content using sensitive or personal information without permission. Addressing these ethical considerations helps ensure that generative AI is used responsibly and benefits both creators and society.


```python
'''
Question 6: Use the following small text dataset to train a simple Variational
Autoencoder (VAE) for text reconstruction:
["The sky is blue", "The sun is bright", "The grass is green",
"The night is dark", "The stars are shining"]
1. Preprocess the data (tokenize and pad the sequences).
2. Build a basic VAE model for text reconstruction.
3. Train the model and show how it reconstructs or generates similar sentences.
Include your code, explanation, and sample outputs.
'''

# Import libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Input, Dense, Layer
from tensorflow.keras.models import Model

# Dataset
sentences = [
    "The sky is blue",
    "The sun is bright",
    "The grass is green",
    "The night is dark",
    "The stars are shining"
]

# Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

sequences = tokenizer.texts_to_sequences(sentences)
word_index = tokenizer.word_index

# Padding
max_length = max(len(seq) for seq in sequences)
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding='post')

vocab_size = len(word_index) + 1

print("Word Index:", word_index)
print("Padded Sequences:\n", padded_sequences)

# Normalize
x_train = padded_sequences.astype("float32") / vocab_size


# Custom Sampling Layer with KL loss
class Sampling(Layer):

    def call(self, inputs):
        z_mean, z_log_var = inputs

        epsilon = tf.random.normal(shape=tf.shape(z_mean))

        z = z_mean + tf.exp(0.5 * z_log_var) * epsilon

        # KL loss added properly inside layer
        kl_loss = -0.5 * tf.reduce_mean(
            tf.reduce_sum(
                1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var),
                axis=1
            )
        )

        self.add_loss(kl_loss)

        return z


# Model parameters
input_dim = max_length
latent_dim = 2
hidden_dim = 8


# Encoder
inputs = Input(shape=(input_dim,))
hidden = Dense(hidden_dim, activation="relu")(inputs)

z_mean = Dense(latent_dim)(hidden)
z_log_var = Dense(latent_dim)(hidden)

z = Sampling()([z_mean, z_log_var])


# Decoder
decoder_hidden = Dense(hidden_dim, activation="relu")(z)
outputs = Dense(input_dim, activation="sigmoid")(decoder_hidden)


# VAE model
vae = Model(inputs, outputs)


# Compile
vae.compile(
    optimizer="adam",
    loss="mse"
)


# Train
vae.fit(
    x_train,
    x_train,
    epochs=300,
    batch_size=2,
    verbose=0
)


# Reconstruction
reconstructed = vae.predict(x_train)

reconstructed_scaled = np.round(reconstructed * vocab_size).astype(int)

reverse_word_index = {v: k for k, v in word_index.items()}


# Decode function
def decode(seq):
    return " ".join(
        reverse_word_index.get(i, "")
        for i in seq if i != 0
    )


print("\nOriginal vs Reconstructed:\n")

for i in range(len(padded_sequences)):

    print("Original:", decode(padded_sequences[i]))

    print("Reconstructed:", decode(reconstructed_scaled[i]))

    print()
```

```python
'''
Question 7: Use a pre-trained GPT model (like GPT-2 or GPT-3) to translate a short
English paragraph into French and German. Provide the original and translated text.
'''

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")

# Original English paragraph
english_text = (
    "Artificial Intelligence is transforming the world by improving healthcare, "
    "education, and business. It helps automate tasks, analyze data, and make "
    "better decisions efficiently."
)

# Function to translate
def translate(text, target_language):

    prompt = f"Translate the following English text to {target_language}: {text}"

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(**inputs, max_length=200)

    return tokenizer.decode(outputs[0], skip_special_tokens=True)


# Translate
french_translation = translate(english_text, "French")
german_translation = translate(english_text, "German")

# Print results
print("Original English Text:\n")
print(english_text)

print("\nFrench Translation:\n")
print(french_translation)

print("\nGerman Translation:\n")
print(german_translation)
```

```python
'''
Question 8: Implement a simple attention-based encoder-decoder model for
English-to-Spanish translation using Tensorflow or PyTorch.
'''

import tensorflow as tf
import numpy as np
from tensorflow.keras.layers import Layer, Embedding, LSTM, Dense, Input, RepeatVector, Concatenate
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Sample dataset
english_sentences = [
    "hello",
    "how are you",
    "i am fine",
    "thank you",
    "good morning"
]

spanish_sentences = [
    "hola",
    "como estas",
    "estoy bien",
    "gracias",
    "buenos dias"
]

# Tokenization
eng_tokenizer = Tokenizer(filters='')
eng_tokenizer.fit_on_texts(english_sentences)
eng_sequences = eng_tokenizer.texts_to_sequences(english_sentences)

spa_tokenizer = Tokenizer(filters='')
spa_tokenizer.fit_on_texts(spanish_sentences)
spa_sequences = spa_tokenizer.texts_to_sequences(spanish_sentences)

# Padding
max_eng_len = max(len(seq) for seq in eng_sequences)
max_spa_len = max(len(seq) for seq in spa_sequences)

encoder_input_data = pad_sequences(eng_sequences, maxlen=max_eng_len, padding='post')
decoder_input_data = pad_sequences(spa_sequences, maxlen=max_spa_len, padding='post')

# Reverse Spanish word index
reverse_spa = {v: k for k, v in spa_tokenizer.word_index.items()}

# Attention Layer (Keras 3 compatible)
class Attention(Layer):

    def __init__(self, units):
        super().__init__()
        self.W1 = Dense(units)
        self.W2 = Dense(units)
        self.V = Dense(1)

    def call(self, encoder_outputs, decoder_hidden):

        decoder_hidden = tf.expand_dims(decoder_hidden, 1)

        score = self.V(
            tf.nn.tanh(
                self.W1(encoder_outputs) +
                self.W2(decoder_hidden)
            )
        )

        attention_weights = tf.nn.softmax(score, axis=1)

        context_vector = attention_weights * encoder_outputs
        context_vector = tf.reduce_sum(context_vector, axis=1)

        return context_vector


# Model parameters
vocab_eng = len(eng_tokenizer.word_index) + 1
vocab_spa = len(spa_tokenizer.word_index) + 1

embedding_dim = 16
units = 32

# Encoder
encoder_inputs = Input(shape=(max_eng_len,))
encoder_embedding = Embedding(vocab_eng, embedding_dim)(encoder_inputs)

encoder_outputs, state_h, state_c = LSTM(
    units,
    return_sequences=True,
    return_state=True
)(encoder_embedding)

# Attention
attention = Attention(units)
context_vector = attention(encoder_outputs, state_h)

context_vector = RepeatVector(max_spa_len)(context_vector)

# Decoder
decoder_inputs = Input(shape=(max_spa_len,))
decoder_embedding = Embedding(vocab_spa, embedding_dim)(decoder_inputs)

decoder_outputs, _, _ = LSTM(
    units,
    return_sequences=True,
    return_state=True
)(decoder_embedding, initial_state=[state_h, state_c])

# Combine context and decoder output
combined = Concatenate()([decoder_outputs, context_vector])

# Output layer
outputs = Dense(vocab_spa, activation="softmax")(combined)

# Build model
model = tf.keras.Model([encoder_inputs, decoder_inputs], outputs)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy"
)

# Train model
model.fit(
    [encoder_input_data, decoder_input_data],
    np.expand_dims(decoder_input_data, -1),
    epochs=300,
    verbose=0
)

# Test translation on full dataset
print("Full Dataset Translation Results:\n")

for eng_sentence in english_sentences:

    test_seq = eng_tokenizer.texts_to_sequences([eng_sentence])
    test_seq = pad_sequences(test_seq, maxlen=max_eng_len, padding='post')

    prediction = model.predict([test_seq, decoder_input_data[:1]], verbose=0)

    predicted_indices = np.argmax(prediction[0], axis=-1)

    translated = " ".join(
        reverse_spa.get(idx, "")
        for idx in predicted_indices
    ).strip()

    print("English:", eng_sentence)
    print("Spanish:", translated)
    print()
```

```python
'''
Question 9: Use the following short poetry dataset to simulate poem generation with a
pre-trained GPT model:
["Roses are red, violets are blue,",
"Sugar is sweet, and so are you.",
"The moon glows bright in silent skies,",
"A bird sings where the soft wind sighs."]
Using this dataset as a reference for poetic structure and language, generate a new 2-4
line poem using a pre-trained GPT model (such as GPT-2). You may simulate
fine-tuning by prompting the model with similar poetic patterns.
Include your code, the prompt used, and the generated poem in your answer.
'''

from transformers import pipeline, set_seed

# Ensure reproducibility
set_seed(42)

# Load pre-trained GPT-2 model
generator = pipeline("text-generation", model="gpt2")

# Poetry dataset
poetry_dataset = [
    "Roses are red, violets are blue,",
    "Sugar is sweet, and so are you.",
    "The moon glows bright in silent skies,",
    "A bird sings where the soft wind sighs."
]

# Create prompt using dataset to simulate fine-tuning
prompt = (
    poetry_dataset[0] + "\n" +
    poetry_dataset[1] + "\n" +
    poetry_dataset[2] + "\n" +
    poetry_dataset[3] + "\n" +
    "Write a new poem in a similar style:\n"
)

# Generate poem
output = generator(
    prompt,
    max_new_tokens=40,
    temperature=0.8,
    do_sample=True,
    pad_token_id=50256
)

# Extract generated poem
generated_poem = output[0]["generated_text"].replace(prompt, "").strip()

# Print prompt and poem
print("Prompt Used:\n")
print(prompt)

print("\nGenerated Poem:\n")
print(generated_poem)
```
